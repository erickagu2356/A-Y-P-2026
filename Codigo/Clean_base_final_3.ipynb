```python
import pandas as pd
import datetime as datetime
import numpy as np
import matplotlib.pyplot as plt
filet = r"/content/drive/MyDrive/Proyecto final/healthcare_dataset.csv"
```

```python
base=pd.read_csv(filet)
```

```python
base
```

```python
base.info()
```

```python
base.isna().sum()
```

```python
base.isnull().sum()
```

```python
print(base.duplicated().sum())
```

```python
base.drop_duplicates(inplace=True)
```

```python
base
```

```python
base["Name"]=base["Name"].str.title()
```

```python
base["Date of Admission"]=pd.to_datetime(base['Date of Admission'])
base["Discharge Date"]=pd.to_datetime(base['Discharge Date'])
base.drop(["Insurance Provider","Billing Amount","Blood Type"],axis=1, inplace=True)
```

```python
base
```

```python
base['Days Hospitalized'] = (base['Discharge Date'] - base['Date of Admission']).dt.days
base
```

```python
limites = [0, 30, 60, 100]
nombres_rangos = ['0 a 30 años', '30 a 60 años', 'Mayores de 60 años']
base['Age Range'] = pd.cut(base['Age'], bins=limites, labels=nombres_rangos)
```

```python
base.drop(["Medication"],axis=1,inplace=True)
```

```python
base.drop(["Doctor","Hospital","Room Number"],axis=1,inplace=True)
```

```python
base.describe()
```

```python
base[base["Age"]<1]
```

```python
conteo_medica = base['Medical Condition'].value_counts()
conteo_medica
```

```python
Conteo_Type_urgent=base['Admission Type'].value_counts()
Conteo_Type_urgent
```

```python
Analisis = base.groupby('Medical Condition')["Days Hospitalized"].mean()
plt.figure(figsize=(8,5))
plt.plot(Analisis.index, Analisis.values)
plt.title("Promedio de dias hospitalizados por tipo de enfermedad")
plt.xlabel("Tipo de enfermedad")
plt.ylabel("Dias hospitalizados")
plt.tight_layout()
plt.show()
```

```python
Grafica2 = base.groupby('Admission Type')["Days Hospitalized"].mean()
plt.figure(figsize=(8,5))
plt.bar(Analisis.index, Analisis.values)
plt.title("Promedio de dias hospitalizados segun el tipo de entrada")
plt.xlabel("Tipo de entrada")
plt.ylabel("Dias hospitalizados")
plt.tight_layout()
plt.show()
```

```python
Type_condition=base["Medical Condition"].value_counts()
plt.figure(figsize=(8,88))
plt.pie(Type_condition.values, labels=Type_condition.index, autopct='%1.1f%%')
plt.title("Porcentaje de internados por tipo de enfermedad")
plt.tight_layout()
plt.show()
```

```python
Age_range=base["Age Range"].value_counts()
plt.figure(figsize=(8,8))
plt.pie(Age_range.values, labels=Age_range.index, autopct='%1.1f%%')
plt.title("Porcentaje de internados por rango de edad")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(base['Test Results'], base['Days Hospitalized'], s=10, alpha=0.3)
plt.title("Resultados del examen vs dias hospitalizados")
plt.xlabel("Resultados del examen")
plt.ylabel("Dias hospitalizados")
plt.tight_layout()
plt.show()
```

```python
Grafica = base.groupby('Age Range')["Days Hospitalized"].mean()
plt.figure(figsize=(8,5))
plt.fill_between(Grafica.index, Grafica.values)
plt.title("Promedio de dias hospitalizados segun el rango de edad")
plt.xlabel("Rango de edad")
plt.ylabel("Dias hospitalizados")
plt.tight_layout()
plt.show()
```

```python
Mayores_60=base[base["Age"]>60]
Entre30_60=base[(base["Age"]<60)& (base["Age"]>30)]
Menor_30=base[base["Age"]<30]
```

```python
M_60=Mayores_60["Medical Condition"].value_counts()
plt.figure(figsize=(8,8))
plt.pie(M_60.values, labels=M_60.index, autopct='%1.1f%%')
plt.title("Distribucion de enfermedad en los mayores de 60")
plt.tight_layout()
plt.show()
```

```python
E_30_60=Entre30_60["Medical Condition"].value_counts()
plt.figure(figsize=(8,8))
plt.pie(E_30_60.values, labels=E_30_60.index, autopct='%1.1f%%')
plt.title("Distribucion de enfermedad en los mayores de 30 y menores de 60")
plt.tight_layout()
plt.show()
```

```python
m_30=Menor_30["Medical Condition"].value_counts()
plt.figure(figsize=(8,8))
plt.pie(m_30.values, labels=m_30.index, autopct='%1.1f%%')
plt.title("Distribucion de enfermedad en los menores de 30")
plt.tight_layout()
plt.show()
```
